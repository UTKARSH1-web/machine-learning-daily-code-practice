```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV
```

```python
data = pd.read_csv(
    "dataset.csv"
)

data.head()
```

```python
X = data.drop(
    "target",
    axis=1
)

y = data["target"]
```

```python
model = DecisionTreeClassifier(
    random_state=42
)
```

```python
param_grid = {
    "criterion": [
        "gini",
        "entropy"
    ],
    "max_depth": [
        2,
        3,
        4,
        5
    ]
}
```

```python
grid_search = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy"
)

grid_search.fit(
    X,
    y
)
```

```python
print(
    "Best Parameters:",
    grid_search.best_params_
)

print(
    "Best Score:",
    grid_search.best_score_
)
```

```python
results = pd.DataFrame(
    grid_search.cv_results_
)

results[
    [
        "params",
        "mean_test_score"
    ]
]
```

```python
plt.plot(
    results["mean_test_score"]
)

plt.title(
    "Grid Search Results"
)

plt.show()
```
