```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import (
    KFold,
    cross_val_score
)
```

```python
data = pd.read_csv(
    "dataset.csv"
)

data.head()
```

```python
X = data.drop(
    "target",
    axis=1
)

y = data["target"]
```

```python
model = DecisionTreeClassifier(
    random_state=42
)
```

```python
kfold = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
scores = cross_val_score(
    model,
    X,
    y,
    cv=kfold,
    scoring="accuracy"
)

scores
```

```python
print(
    "Average Accuracy:",
    scores.mean()
)
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    range(1,6),
    scores
)

plt.xlabel("Fold")

plt.ylabel("Accuracy")

plt.title(
    "K-Fold Scores"
)

plt.ylim(0,1)

plt.show()
```
