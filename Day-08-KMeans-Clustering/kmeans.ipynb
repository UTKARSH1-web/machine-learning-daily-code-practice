```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
```

```python
data = pd.read_csv("dataset.csv")

X = data[["x", "y"]]
```

```python
model = KMeans(
    n_clusters=4,
    random_state=42
)
```

```python
model.fit(X)
```

```python
labels = model.labels_

centers = model.cluster_centers_
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    data["x"],
    data["y"],
    c=labels
)

plt.scatter(
    centers[:, 0],
    centers[:, 1],
    marker='X',
    s=200
)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("K-Means Clustering")

plt.show()
```
