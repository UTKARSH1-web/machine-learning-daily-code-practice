```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import LabelEncoder
```

```python
data = pd.read_csv("dataset.csv")

data.head()
```

```python
encoder = LabelEncoder()

data["gender"] = encoder.fit_transform(
    data["gender"]
)

data["city"] = encoder.fit_transform(
    data["city"]
)

data["purchased"] = encoder.fit_transform(
    data["purchased"]
)

data
```

```python
data.info()
```

```python
data["city"].value_counts()
```

```python
data["city"].value_counts().plot(
    kind="bar"
)

plt.title(
    "Encoded City Distribution"
)

plt.show()
```
