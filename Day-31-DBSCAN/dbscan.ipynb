```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import DBSCAN
```

```python
data = pd.read_csv(
    "dataset.csv"
)

data.head()
```

```python
X = data[
    ["feature1", "feature2"]
]
```

```python
dbscan = DBSCAN(
    eps=1.0,
    min_samples=3
)

clusters = dbscan.fit_predict(X)

clusters
```

```python
data["cluster"] = clusters

data
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    data["feature1"],
    data["feature2"],
    c=data["cluster"]
)

plt.title(
    "DBSCAN Clusters"
)

plt.show()
```
